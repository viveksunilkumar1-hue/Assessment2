```python
import numpy as np
import pandas as pd
```

```python
#Q1. NumPy Array Creation – 2 Marks
#Create a NumPy array containing the following values:
arr=np.array([10,20,30,40,50])
print(arr)
print(arr.shape)
print(arr.dtype)
```

```python
#Q2. NumPy Array Operations – 2 Marks


arr=np.array([5,10,15,20,25])
print(arr+5)
print(arr*2)
```

```python
#Q3. NumPy Statistical Operations – 2 Marks

arr= np.array([12,15,18,20,25,30])
print("mean:",arr.mean())
print("median:",np.median(arr))
print("Minimum:",arr.min())
print("Maximum:",arr.max())
print("Standard Deviation:",arr.std())


```

```python
#Q4. NumPy Filtering – 2 Marks
#Given:
#arr = np.array([10, 25, 30, 45, 50, 65, 70])
#Using NumPy, extract all values that are greater than 40.

arr = np.array([10, 25, 30, 45, 50, 65, 70])
arr[arr>40]
```

```python
#Q10. Handling Missing Values – 2 Marks
#Consider the following DataFrame:


df = pd.DataFrame({
'Name': ['A', 'B', 'C', 'D'],
'Age': [21, np.nan, 23, 22],
'Score': [80, 75, np.nan, 90]})
print(df.isnull())
print(df['Age'].fillna(df['Age'].mean()))
print(df['Score'].fillna(df['Score'].mean()))


```

```python
#Q11. Data Loading and Exploration – 6 Marks
#Load student_performance.csv into a Pandas DataFrame.


file_path='/content/student_performance.csv'
df=pd.read_csv(file_path)
print(df.head())
print(df.tail())
print(df.shape)
print(df.columns)
print(df.dtypes)
print(df.describe)



```

```python
#Q12. Filtering and Sorting – 6 Marks


file_path='/content/student_performance.csv'
df=pd.read_csv(file_path)
print(df[df['Final_Score']>75])
print(df[df['Attendance']<75])
print(df[(df['Department']=='Data Science')&(df['Final_Score']>80)])
print(df.sort_values(by='Final_Score',ascending=False))

```

```python
#Q13. Feature Creation – 6 Marks

df['Average_Technical_Score']=(df['Python_Score']+df['SQL_Score']+df['Pandas_Score']+df['NumPy_Score'])
print(df)
```

```python
#Q14. GroupBy Analysis – 6 Marks

file_path='/content/student_performance.csv'
df=pd.read_csv(file_path)
print(df.groupby('Department').agg({'Final_Score':['mean','max','min']}))
print(df.sort_values(by='Final_Score',ascending=False))






```
